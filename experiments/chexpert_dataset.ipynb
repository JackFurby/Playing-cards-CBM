```python
import pandas as pd
```

```python
file_path = '../dataset/chexpert/train_cheXbert.csv'

data = pd.read_csv(file_path)

# Replace values
data = data.fillna(0)
data = data.replace(-1.0, 1)

print(data.head())
```

# Dataframe stats

```python
total_rows = len(data)

numeric_columns = data.select_dtypes(include=[int, float]).columns
column_sums = data[numeric_columns].sum()

print(f"Total number of rows: {total_rows}")

print("\nSum of columns:")
print(column_sums)
```

```python
filtered_data_findings = data[data.iloc[:, -1] == 1]

concept_columns = filtered_data_findings.iloc[:, -14:-1] # get all columns for concepts

# Get the count of unique rows and their occurrences for the last 14 columns
unique_rows_counts = concept_columns.apply(lambda row: tuple(row), axis=1).value_counts()

for row, count in unique_rows_counts.items():
    print(f"{row}- Sum {sum(row)} - Occurrences: {count}")
    print([idx if i == 1.0 else "_" for idx, i in enumerate(row)])
```

```python
filtered_data_no_findings = data[data.iloc[:, -1] == 0]

concept_columns = filtered_data_no_findings.iloc[:, -14:-1] # get all columns for concepts

# Get the count of unique rows and their occurrences for the last 14 columns
unique_rows_counts = concept_columns.apply(lambda row: tuple(row), axis=1).value_counts()

for row, count in unique_rows_counts.items():
    print(f"{row}- Sum {sum(row)} - Occurrences: {count}")
    print([idx if i == 1.0 else "_" for idx, i in enumerate(row)])
```
